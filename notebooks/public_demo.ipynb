# GEE Workbench — public demo

This notebook uses public Earth Engine datasets only. Configure your own Cloud project before running.

```python
from pathlib import Path
import pandas as pd
import ee
import geemap
from gee_workbench.config import load_config
from gee_workbench.manager import doctor, concise_assets, concise_tasks, list_assets, list_tasks

config = load_config()
ee.Initialize(project=config.project)
print(f'Initialized project: {config.project}')
```

## 1. Project health

```python
health = doctor(config)
pd.DataFrame({
    'project': [health['project']],
    'asset_count': [health['assets']['total']],
    'task_count': [health['tasks']['total']],
    'earthengine_api': [health['earthengine_api']],
})
```

## 2. Read-only asset and task inventory

```python
assets_df = pd.DataFrame(concise_assets(list_assets(config)))
tasks_df = pd.DataFrame(concise_tasks(list_tasks()))
display(assets_df)
display(tasks_df)
```

## 3. Interactive public-data map

SRTM elevation and JRC Global Surface Water are public catalog datasets.

```python
elevation = ee.Image('USGS/SRTMGL1_003')
water = ee.Image('JRC/GSW1_4/GlobalSurfaceWater').select('occurrence')

Map = geemap.Map(center=[20, 0], zoom=2)
Map.add_basemap('OpenStreetMap')
Map.addLayer(elevation, {'min': 0, 'max': 4000, 'palette': ['0B3D2E', '7CB342', 'E6D690', '8D6E63', 'FFFFFF']}, 'SRTM elevation', False)
Map.addLayer(water, {'min': 0, 'max': 100, 'palette': ['FFFFFF', 'A9D6E5', '0077B6', '023E8A']}, 'JRC water occurrence', False)
Map.add_layer_manager()
Map
```
